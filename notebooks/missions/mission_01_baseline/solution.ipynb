# Mission 01: Initial LangChain Agent (Baseline Chatbot) - 해답 노트북

이 노트북은 첫 번째 미션의 완성된 솔루션 코드와 설명입니다.

```python
# 1. 필요한 라이브러리 및 환경 로드
import sys
import os
from dotenv import load_dotenv

while not os.path.exists("app") and os.getcwd() != "/":
    os.chdir("..")
# 루트 폴더 기준의 경로 등록
sys.path.append(os.path.abspath("src"))
sys.path.append(os.path.abspath("."))  # sys.path.append("app") 대신 "." 등록이 파이썬 패키지 경로 탐색에 안전합니다.
load_dotenv(override=True)

from app.utils.llm import get_llm
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
```

### [미션 1] 에이전트 인스턴스 완성하기

아래 코드는 `get_llm`과 `create_agent`를 활용하여 에이전트를 완성한 솔루션입니다.

```python
# LLM 로드 (gemini-3.5-flash 모델 적용)
llm = get_llm(model_name="gemini-3.5-flash", temperature=0.0)

# 에이전트 생성
from app.prompts import CHATBOT_SYSTEM_PROMPT
from app.utils.context import AgentContext
from langgraph.checkpoint.memory import MemorySaver

memory = MemorySaver()
agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=CHATBOT_SYSTEM_PROMPT,
    checkpointer=memory,
    context_schema=AgentContext
)

if agent is not None:
    print("✅ 에이전트 생성 성공!")
else:
    print("❌ 에이전트 생성 실패!")
```

### [미션 2] 에이전트 호출 검증

에이전트의 대화 루프를 정상 호출하여 작동 여부를 테스트합니다.

```python
if agent:
    res = agent.invoke(
        {"messages": [HumanMessage(content="안녕하세요! 오늘 하루는 어때요?")]},
        config={"configurable": {"thread_id": "test_thread_01"}}
    )
    print("답변:", res["messages"][-1].content)
else:
    print("에이전트가 존재하지 않습니다.")
```
