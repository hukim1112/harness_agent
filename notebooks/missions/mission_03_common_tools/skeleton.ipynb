### [미션 3] 자율 실행 루프(ReAct) 검증

에이전트가 질문을 해결하기 위해 도구를 자율적으로 호출하는지 실행 로그를 통해 테스트합니다.

```python
if agent:
    # 1. 파일 검색 및 읽기 복합 태스크 테스트
    # LLM이 glob_search로 파일들을 확인한 후 file_read로 세부 내용을 읽어들일 것입니다.
    res1 = agent.invoke({"messages": [HumanMessage(content="app 폴더 아래 server.py 파일의 1~15줄 내용을 읽어서 보여줘.")]})
    print("답변1:", res1["messages"][-1].content)
    
    print("\n" + "="*50 + "\n")
    
    # 2. 시스템 명령어 실행 테스트
    # LLM이 bash_command를 골라 셸 명령을 실행할 것입니다.
    res2 = agent.invoke({"messages": [HumanMessage(content="현재 리눅스 터미널의 파이썬 버전을 bash_command로 확인해서 요약해줘.")]})
    print("답변2:", res2["messages"][-1].content)
else:
    print("에이전트가 준비되지 않았습니다.")
```
