## Step 1 - Using a `for` loop

We can initialize a variable `total` to 0, and then iterate through numbers from 1 to 100, adding each number to `total`.

```python
# Calculate sum using a for loop
total = 0
for i in range(1, 101):
    total += i

print(f"The sum of numbers from 1 to 100 using a loop is: {total}")
```

## Step 2 - Using the built-in `sum()` and `range()` functions

Python provides a built-in `sum()` function that can sum up elements in an iterable, such as a `range` object.

```python
# Calculate sum using sum() and range()
total_builtin = sum(range(1, 101))
print(f"The sum of numbers from 1 to 100 using sum() is: {total_builtin}")
```

## Step 3 - Using the mathematical formula (Arithmetic Progression)

The sum of numbers from 1 to $n$ can be calculated using the formula:
$$\text{Sum} = \frac{n(n + 1)}{2}$$

For $n = 100$:
$$\text{Sum} = \frac{100 \times 101}{2} = 5050$$

```python
# Calculate sum using the mathematical formula
n = 100
total_formula = n * (n + 1) // 2
print(f"The sum of numbers from 1 to 100 using the formula is: {total_formula}")
```
